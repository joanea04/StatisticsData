Slide 3 - Exercise 2: Catalog Analysis and Filtering
What to say

This second exercise was all about turning a huge, messy catalog into a clean, trustworthy dataset we could actually train a model on.

We started with the joined Galaxy Zoo and SDSS catalog — about 659,000 objects across 90 columns. Clearly far more than we needed, and not all of it reliable.

Our first cut was on morphology: we kept only the galaxies confidently classified as spiral or elliptical, and dropped everything labeled 'uncertain.' That alone took us down to about 250,000 objects.

Next we looked at photometry quality. We worked in the g, r, i, and z bands, removed the invalid placeholder values and any measurements with errors above 0.1 magnitudes, and we deliberately dropped the u band because it was far too noisy — and importantly, requiring good u-band data would have thrown out ellipticals at a much higher rate, biasing our sample. After this cut we still had around 249,700 galaxies, so photometry quality barely cost us anything.

Then we removed duplicate objects, keeping the closest match, which left us with our final sample of 249,155 unique galaxies — about 38% of what we started with.

Finally, we engineered our actual features: the three colour indices, g minus r, r minus i, and i minus z.

And this plot at the bottom is really the payoff. You can see the ellipticals in red sit at redder colours than the spirals in blue, with the clearest separation in g minus r — which is exactly the physics we expected, since ellipticals have older, redder stellar populations. That told us these colours carry real information for the classifier.

Slide 5 - Exercise 3: Image Cutout Retrieval
What to say

"The second half of the exercise was about building an actual image dataset of these galaxies.

Our goal here was specific: build a balanced image dataset with an equal number of spirals and ellipticals. We ended up targeting seven thousand successfully downloaded images per class, so fourteen thousand galaxies in total, evenly split.

To get the images we used the CDS HiPS2FITS service. For each galaxy we took its coordinates from our clean catalog and requested a cutout centered on it.
Now, we made a few decisions to check everything was running correctly.

First, validation. Before accepting any image, we checked that it actually opened, that it was a real JPEG, and that it had the exact dimensions we expected. That stops corrupted or empty downloads from silently polluting the dataset.

Second, reliability. Each download retries up to three times with a short pause in between before giving up. And importantly, we didn't pre-pick exactly seven thousand galaxies — we kept the full candidate pool, shuffled with a fixed random seed for reproducibility, and simply drew the next galaxy whenever one failed. So we always reached our target count of valid images.

We also downloaded in parallel using a small pool of four workers — fast enough to save a lot of time, but deliberately limited so we didn't hammer a shared public service.

The Images were saved into folders by class, every attempt is logged, and each image links back to its catalog entry through the galaxy's object ID. A nice practical benefit of logging and saving incrementally is that if we rerun it, it keeps the images we already have and just fills in the rest — so we could start small with a dozen images to check quality, then scale up to thousands without redoing work.

The end result was a clean, balanced, validated set of fourteen thousand galaxy images — spirals and ellipticals in equal numbers — ready to train the classifier in the final exercise."

Slide 6 - Exercise 4: Programming Topic Introspection
For Exercise 4, we researched two important software-development practices: testing and documentation.

"Exercise four stepped back from the science to look at good programming practice — specifically testing and documenting.

On testing: Testing is the practice of checking that code does what you intend — and keeps doing it as the code changes. The underlying philosophy is that you shouldn't just trust code because it "ran without an error"; you actively verify its behaviour against expectations. Ee verified our filtered catalog only contained valid galaxies, checked that the morphology labels were assigned correctly, confirmed the coordinates were valid before fetching images, made sure the downloaded cutouts had the right dimensions, and did spot visual inspections of random galaxies.

On documenting: Documentation is written explanation that lets someone — including your future self — understand, use, and maintain code without having to reverse-engineer it. The key insight is that code tells you how something works, but documentation tells you what it does and why. Good documentation takes forms like code comments, READMEs, technical reports, and notebook explanations. In our project we documented our data sources, explained our filtering criteria, recorded the image-retrieval parameters, and kept our notebooks organized into clear, explained sections.

In our project, documentation was especially important because the workflow contains multiple stages, from catalog filtering to image retrieval. We documented the data sources, filtering rules, and processing steps, and structured our notebooks so that each stage of the pipeline could be easily understood and reproduced."
